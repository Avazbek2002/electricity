```python
import pandas as pd
d = pd.read_csv("transformer_daily.csv", parse_dates=["calendar_day"])
day = d.groupby("calendar_day").agg(cov=("coverage", "median"), rep=("n_reporting", "sum"),
                                    ever=("n_ever", "sum"))
day["cov_all"] = day["rep"] / day["ever"]
low = day[day["cov_all"] < 0.8]
print(low["cov_all"].round(2).to_string())          # which days, and how low
```

```python
import pandas as pd
d = pd.read_csv("transformer_daily.csv", parse_dates=["calendar_day"])
d["tid"] = d["transformer_id"].astype(str)
t = d[d["tid"] == "862928"].set_index("calendar_day")
gap = t.index[t["coverage"] < 0.8]

# on 862928's gap days, what share of the OTHER transformers is also below 0.8?
o = d[d["calendar_day"].isin(gap) & (d["tid"] != "862928")]
print(o.groupby("calendar_day")["coverage"].apply(lambda s: (s < 0.8).mean()).describe())

# how far below the threshold is it on those days?
print(t.loc[gap, ["n_reporting", "n_ever", "coverage"]].describe())

# which transformers have the most low-coverage days (the system-wide outages are in every row)
bad = d.assign(low=d["coverage"] < 0.8).groupby("tid").agg(
    low_days=("low", "sum"), n_ever=("n_ever", "max")).sort_values("low_days", ascending=False)
print(bad.head(15))
```
